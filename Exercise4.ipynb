#PROGRAM THAT COUNTS NUMBER OF VOWELS IN A STRING

```python
n=input()
count=0
for i in n:
  if i=='a' or i=='A' or i=='e' or i=='E' or i=='I' or i=='i' or i=='o' or i=='O' or i=='u' or i=='U':
    count+=1
print(count)
```

#PROGRAM THAT CHECKS SUB STRING IS PRESENT IN THE GIVEN STRING, IF PRESENT PRINT THE INDEX

```python
s1=str(input("Enter a string:"))
s2=str(input("Enter a sub string:"))
if s2 in s1:
  print(s1.find(s2))
else:
  print("NO")
```

#PROGRAM THAT PRINTS REVERSE OF THE NUMBER USING FUNCTIONS

```python
def reverse(a):
  rev=0
  while(a!=0):
    rem=a%10
    rev=rev*10+rem
    a=a//10
  return rev
a=int(input())
print(reverse(a))
```
