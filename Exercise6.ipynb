#TUPLE OPERATIONS

```python
t=(1,2,"name")
t1=(3,4,"maha")
#concatenation
print(t+t1)
#repetition
print(t1*2)
#indexing
print(t1[2])
#membership operator
print(3 in t1)
#iteration
for i in t1:
  print(i)
```

#BUILT-IN FUNCTIONS

```python
t1=tuple((3,4,5,6,7))
#sum
print(sum(t1))
#len
print(len(t1))
#min
print(min(t1))
#max
print(max(t1))
#reversed
print(reversed(t1))
#sorted
print(sorted(t1))
```

#COUNT, INDEX

```python
t1=(1,2,3,4,5)
print(t1.count(2))
print(t1.index(3))
```

#ZIP FUNCTION

```python
t1=(1,2,3,"nissy")
t2=(4,5,7,"maha")
print(tuple(zip(t1,t2)))
```

#WRITE A PROGRAM TO COUNT THE NUMBER OF VOWELS IN A STRING (NO CONTROL FLOW ALLOWED)

```python
s1=input("Enter a string:")
s2=s1.lower()
vowel_count=s2.count("a")+s2.count("e")+s2.count("i")+s2.count("o")+s2.count("u")
print("The number of vowels in the string are:",vowel_count)
```
