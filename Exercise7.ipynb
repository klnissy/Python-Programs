#SETS IN PYTHON

#SET METHODS

```python
A={1,2,4,6,8,10}
B={1,2,3,4,5,6}
#add->Add one element
A.add(7)
print(A)
#update->Add multiple elements
A.update([11,13])
print(A)
#remove->Remove element; error if missing
A.remove(6)
print(A)
#discard->Remove element; no error if missing
A.discard(10)
print(A)
#pop->Remove and return an arbitrary element
A.pop()
print(A)
#clear->Remove all elements
A.clear()
print(A)
#union->Elements from either set
print(A.union(B))
#intersection->Elements common to both
print(A.intersection(B))
#difference->Elements in a but not b
print(A.difference(B))
print(B.difference(A))
#symmetric_difference->Elements in either, but not both
print(A.symmetric_difference(B))
```

#WRITE A PYTHON PROGRAM TO FIND HOW MANY NUMBERS ARE HAVING DUPLICATES IN A LIST

```python
L=[int(i) for i in input().split()]
L2=[]
for i in L:
  if L.count(i)>1:
    L2.append(i)
print(L2)
s=set(L2)
print(s)
print(len(s))
```
