This code is an example of arithmetic operators

```python
a = int(input("Enter a value:"))
b = int(input("Enter another value:"))
print("Addition:", a + b)
print("Subtraction:", a - b)
print("Multiplication", a * b)
print("Division:", a / b)
print("Floor Division:", a // b)
print("Modulus:", a % b)
print("Exponent:", a ** b)
print("Adding", a, "to", b, "gives", a + b)
```

This code is an example of comparision operators

```python
c = int(input("Enter a value:"))
d = int(input("Enter another value:"))
print(c > d)
print(c < d)
print(c >= d)
print(c <= d)
print(c == d)
print(c != d)
```

This code is an example of logical operators

```python
e = int(input("Enter a value:"))
f = int(input("Enter another value:"))
print(e < 7 and f > 7)
print(e < 7 or f > 7)
print(not(e < 7 and f > 7))
print(not(e < 7 or f > 7))
```
