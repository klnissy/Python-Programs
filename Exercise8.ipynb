#PROGRAM THAT PRINTS FREQUENCY OF EACH ELEMENT IN A LIST IN DICT FORMAT
#FORMAT [1,2,3,4,1,5,6,1,6,7]  {1:3,2:1,3:1,4:1,5:1,6:2,7:1}

```python
L=list(map(int,input().split()))
d={}
for i in L:
  d[i]=L.count(i)
print(d)
```
