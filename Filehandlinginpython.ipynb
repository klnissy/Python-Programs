```python
#program on w mode
file=open("f1.txt","w")
data="hello world"
file.write(data)
print("file created")
file.close()
```

```python
#program on r mode
with open("f1.txt","r") as f:
  data=f.read()
  print(data)
```

```python
with open("f1.txt","a") as f:
  f.writelines(["\napple","\nmango","\nend"])
```

```python
with open("f1.txt","r") as f:
  print(f.readline())
  print(f.read())
  print(f.readlines())
```

```python
with open("f1.txt","r") as f:
  print(f.readline())
  print(f.tell())
  f.seek(1)
  print(f.tell())
  print(f.read())
```

```python
#Program that copies content from one file into another file
with open("new.txt","r") as f:
  data=f.read()
with open("f2.txt","w") as f1:
  f1.write(data)
print("File copied")

```

```python
#Program to print each line in a file in reverse order
with open("f1.txt","r") as f:
  data=f.readlines()
  for i in data:
    print(i[::-1])
```

```python
#Program to sort data in a file and put them in another file.The output file should have only lower-case words so any upper-case words from source file must be lowered
with open("new.txt","r") as f:
  data=f.read().split()
  data_lower=[i.lower() for i in data]
  sort_data=sorted(data_lower)
with open("f3.txt","w") as f1:
  for i in sort_data:
    f1.write(i+"\n")
print("done")
```

```python
#Python program to compute the  number of characters,words and lines in a file
with open("new.txt","r") as f:
  data = f.read()
  characters = len(data)
  words = len(data.split())
  lines = len(data.splitlines())
  print("Number of characters:", characters)
  print("Number of words:", words)
  print("Number of lines:", lines)
```

```python
#rb,wb,ab
with open("pic.jpg","rb") as f:
  data=f.read()
with open("pic1.jpg","wb") as f1:
  f1.write(data)
print("pic created")
```

```python
import pickle
with open("s1.dat","wb") as f:
  data=[1,2,3,"hello"]
  pickle.dump(data,f)
with open("s1.dat","rb") as f2:
  d=pickle.load(f2)
  print(d)
```

```python
import pickle
with open("s2.dat","wb") as f:
  data=dict(name="Maha",age=18)
  pickle.dump(data,f)
with open("s2.dat","rb") as f1:
  d=pickle.load(f1)
  print(d)
```
