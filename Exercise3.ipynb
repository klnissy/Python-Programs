#List operations

```python
L1=[1,2,3,"Nissy"]
L2=[5,6,7,"Ma"]
#1 CONCATENATION
print(L1+L2)
#2 REPETITION
print(L1*2)
#3 INDEXING
print(L1[0])
#4 SLICING
print(L1[1:3])
#5 MEMBERSHIP OPERATOR
print(3 in L1)
#6 ITERATION
for i in L1:
  print(i)
```

Built-in Functions

```python
L3=[3,5,7]
#sum()
print(sum(L3))
#len()
print(len(L3))
#min()
print(min(L3))
#max()
print(max(L3))
#REVERSED
print(reversed(L3))
#sorted()
print(sorted(L3))
```

Built-in Methods

```python
L4=[2,4,6,8]
#APPEND
L4.append(5)
print(L4)
#EXTEND
L4.extend([7,9])
print(L4)
#INSERT
L4.insert(1,3)
print(L4)
#POOP
L4.pop()
print(L4)
#REMOVE
L4.remove(6)
print(L4)
#INDEX
print(L4.index(7))
#COUNT
print(L4.count(7))
#REVERSE
L4.reverse()
print(L4)
#SORT
L4.sort()
print(L4)
```

#1 Program that prints sum of list elements without using sum()

```python
L5=[1,2,3,4,5]
s=0
for i in L5:
  s+=i
print(s)
```

#2 Print the list elements at odd index

```python
for i in range(len(L5)):
  if(i%2!=0):
    print(L5[i])
```

#3 Print the even elements

```python
for i in L5:
  if(i%2==0):
    print(i)
```

#List Comprehension

```python
n=int(input())
L=[]
for i in range(n):
  element=int(input())
  L.append(element)
print(L)
```

```python
a=list(map(int,input().split()))
print(a)
```

```python
#List Comprehension
L=[i**2 for i in range(4)]
print(L)
L2=[i**3 for i in L]
print(L2)
L3=[i for i in L2 if i%2==0]
print(L3)
L4=[int(i) for i in input().split()]
print(L4)
```
